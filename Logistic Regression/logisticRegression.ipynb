```python
from sklearn.datasets import load_breast_cancer
```

```python
import numpy as np
```

```python
# Dataset
X,Y = load_breast_cancer(return_X_y=True)
n_samples, n_features = X.shape
print("samples : ", n_samples, " and features : ", n_features)
```

```python
print(X[:5])
```

```python
#normalize X
X = (X - X.mean(axis=0)) / X.std(axis=0)
print(X[:5])
```

```python
#add bias term
X = np.hstack([np.ones((n_samples,1)),X])
print(X[:2])
```

```python
X_train, X_test = X[:400], X[400:]
Y_train, Y_test = Y[:400], Y[400:]
```

```python
print(Y[:2])
```

```python
def sigmoid(z):
    return 1.0/(1.0+np.exp(-z))
```

```python
def accuracy_score(y_true, y_pred):
    return np.mean(y_true == y_pred)*100
```

```python
def calculate_gradient(theta, X, Y_actual):
  
    n_samples = X.shape[0]
    Y_pred = np.zeros(n_samples)

    for i in range(n_samples):
        z = np.dot(theta, X[i])
        y_pred = sigmoid(z)
        # not converting to binary class labels, we will use the predicted probabilities to calculate the gradient
        # as probabilities will provide more information about the confidence of the predictions, which can lead to better gradient estimates and faster convergence during training.
        # during prediction we will convert to binary class labels using the threshold
        Y_pred[i] =y_pred

    print("loss", -np.mean(Y_actual * np.log(Y_pred) + (1 - Y_actual) * np.log(1 - Y_pred)))
    return (1/n_samples) * X.T.dot(Y_pred - Y_actual)

"""
    Y = (n,)
    X = (n,d)
    theta = (d,)
    X.T = (d,n)
    X.T.dot(Y_pred - Y_actual) = (d,n).(n,) = (d,)

    and then we scalet it by 1/n_samples to get the average gradient.
"""
```

```python
def gradient_descent(X,Y , alpha, epochs,tol= 1e-7):

    print("Starting Gradient Descent...")
    theta = np.zeros(X.shape[1])  #initialize theta with zeros
    # theta size same as number of features + 1 for bias term   

    for epoch in range(epochs):
        gradient = calculate_gradient(theta,X,Y)

        #is change is too small, we can stop
        if(np.linalg.norm(gradient) < tol):
            print(f"Convergence reached at epoch {epoch}")
            break
    
        theta = theta - alpha*gradient

    return theta
```

```python
def fit(X,Y,alpha=0.01, epochs=1000):
   return gradient_descent(X,Y,alpha,epochs,tol=1e-7)
```

```python
optimized_theta = fit(X_train,Y_train)
```

```python
print("optimized theta : ", optimized_theta)
```

```python
def predict(X,theta,threshold=0.5):
    n_samples = X.shape[0]
    Y_pred = np.zeros(n_samples)

    for i in range(n_samples):
        z = np.dot(theta, X[i])
        y_pred = sigmoid(z)
        Y_pred[i] = 1 if y_pred >= threshold else 0

    return Y_pred
```

```python
def accuracy(Y_actual, Y_pred):
    return np.mean(Y_actual == Y_pred)*100
```

```python
predictions = predict(X_test, optimized_theta)  
print("Predictions : ", predictions)

test_accuracy = accuracy(Y_test, predictions)
print("Test Accuracy : ", test_accuracy)
```
