```python
import pandas as pd
```

```python
APP_USAGE_TRACE_COLUMNS = [
    "UserID",
    "Timestamp",
    "AppID",
    "BaseID",
    "Traffic",
]

APP2CATEGORY_COLUMNS = [
    "AppID",
    "CategoryID",
]

# base_poi no change as it already has the column names we want

CATEGORYS=[
    "CategoryID",
    "EnglishName",
]
```

```python
def load_add_column_names(filepath,column_names):
    df = pd.read_csv(filepath)
    df.columns = column_names

    #saving it back
    df.to_csv(filepath,index=False)

    print(df.head())
    print(df.columns)

```

```python
load_add_column_names("./dataset/App_usage_trace/app_usage_trace.csv",APP_USAGE_TRACE_COLUMNS)
load_add_column_names("./dataset/App2Category/app2category.csv",APP2CATEGORY_COLUMNS)
load_add_column_names("./dataset/Categorys/categorys.csv",CATEGORYS)
```

```python
#removing the columns that are not needed in the base_poi dataset
filepath="./dataset/base_poi/base_poi.csv"
df=pd.read_csv(filepath)

df.columns= df.iloc[0] # set the first row as column names
df=df[1:] # remove the first row which is now the header

df=df.reset_index(drop=True) # reset the index after removing the first row

df.to_csv(filepath,index=False)
```
