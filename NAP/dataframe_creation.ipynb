```python
""" 
since this data is in str format we need to convert it into numeric form
we will convert to dataframe and then we will convert the data into numeric form 

Dataframe is a 2-dimensional labeled data structure with columns of potentially different types. 
It is similar to a spreadsheet or SQL table, and is a common data structure used in data analysis and manipulation.
"""

def convert_to_dataframe(filepath):
    data = []
    with open(filepath,'r') as file:
        for line in file:
            # parts = line.strip().split() # for app_usage_trace file
            parts = line.strip().split('\t') # for rest files
            # strip() removes leading and trailing whitespace, including the newline character.
            # split() splits the line into parts based on whitespace.
            data.append(parts)


    return pd.DataFrame(data)
```

```python
# creating dataframes for each file and saving them as csv files

df_app_usage_trace = convert_to_dataframe(r'D:\Courses\ml\NAP\dataset\App_usage_trace\App_usage_trace.txt')
print(type(df_app_usage_trace))

#saving the dataframe to csv file
df_app_usage_trace.to_csv(r'D:\Courses\ml\NAP\dataset\App_usage_trace\App_usage_trace.csv', index=False)

df_app2category = convert_to_dataframe(r'D:\Courses\ml\NAP\dataset\App2Category\App2Category.txt')
print(type(df_app2category))
df_app2category.to_csv(r'D:\Courses\ml\NAP\dataset\App2Category\App2Category.csv', index=False)


df_base_poi = convert_to_dataframe(r'D:\Courses\ml\NAP\dataset\base_poi\base_poi.txt')
print(type(df_base_poi))
df_base_poi.to_csv(r'D:\Courses\ml\NAP\dataset\base_poi\base_poi.csv', index=False)


df_Categorys = convert_to_dataframe(r'D:\Courses\ml\NAP\dataset\Categorys\Categorys.txt')
print(type(df_Categorys))
df_Categorys.to_csv(r'D:\Courses\ml\NAP\dataset\Categorys\Categorys.csv', index=False)
```
